```python
import pandas as pd
import pymongo
```

```python
from dotenv import load_dotenv
import os

load_dotenv("../.env", override=True)
```

```python
df = pd.read_csv("data.csv")
df.head(5)
```

```python
len(df)
print("Total number of records in the dataset: ", len(df))
```

```python
# convert the data into dict coz mongoDB only accepts data in dict format
data = df.to_dict(orient="records")

data[0]
```

```python
DB_NAME = "Proj1"
DB_COLLECTION_NAME = "Proj1-Data"
MONGODB_CONNECTION_STRING = os.getenv("MONGODB_CONNECTION_STRING")
 
```

```python
client = pymongo.MongoClient(MONGODB_CONNECTION_STRING)
print(client.admin.command("ping"))
database = client[DB_NAME]
collection = database[DB_COLLECTION_NAME]
```

```python
collection.delete_many({})
```

```python
print(collection.count_documents({}))
```

```python
collection.insert_many(data)
```

```python
print("Total number of records in the collection: ", collection.count_documents({}))
```
